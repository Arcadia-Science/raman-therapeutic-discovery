```python
from pathlib import Path
from plotly.subplots import make_subplots
import arcadia_pycolor as apc
import ramanspy as rp

from raman_analysis.utils import collect_spectra_from_dir, plot_spectra_in_dir

%load_ext autoreload
%autoreload 2

apc.plotly.setup()

# Data directory relative to notebook
parent_dir = Path('../data')

fig_h, fig_w = 800, 1000
spectrum_lims = (700, 3200)
y_offset = (1000, 5000)

pipeline = rp.preprocessing.Pipeline(
    [
        rp.preprocessing.misc.Cropper(region=spectrum_lims),
        rp.preprocessing.despike.WhitakerHayes(),
        rp.preprocessing.denoise.SavGol(window_length=5, polyorder=3),
        # rp.preprocessing.baseline.ASPLS(lam=1e4),
        rp.preprocessing.baseline.ModPoly(poly_order=5),
        # rp.preprocessing.normalise.AUC(),
    ]
)
```

### Wasatch 532 on 2025-07-24

```python
data_dir_sers = parent_dir / 'Trove-SERS-attempt3' / '2025-07-24'

data_list_sers, sample_df_sers = collect_spectra_from_dir(data_dir_sers, 'wasatch')

n_groups = sample_df_sers['Group ID'].nunique()

sampled_custom_colors = ["gray", "red"]

data_list_sers = [d for d in data_list_sers if "20mw" in d[0]]

f, names, rp_container = plot_spectra_in_dir(data_list_sers,
                    col_to_plot = "Intensity",
                    data_dir = data_dir_sers,
                    rp_pipeline=pipeline,
                    colors=sampled_custom_colors,
                    height=fig_h,
                    width=fig_w,
                    title='Raman on Whole extract with Wasatch 785',
                    xaxis_title='Wavenumber (cm⁻¹)',
                    yaxis_title='Intensity (a.u.)',
                    y_offset=y_offset[1]
)
# add_peak_annotations(f)
# f.show()


data_dir_whole = parent_dir / '2025-03-18-ticks'

data_list_whole, sample_df_whole = collect_spectra_from_dir(data_dir_whole, 'wasatch')

# data_list_whole = [d for d in data_list_whole if "20mw" in d[0]]

n_groups = sample_df_whole['Group ID'].nunique()

f2, names, rp_container = plot_spectra_in_dir(data_list_whole,
                    col_to_plot = "Intensity",
                    data_dir = data_dir_whole,
                    rp_pipeline=pipeline,
                    colors=sampled_custom_colors,
                    height=fig_h,
                    width=fig_w,
                    title='SERS on fraction with Wasatch 532',
                    xaxis_title='Wavenumber (cm⁻¹)',
                    yaxis_title='Intensity (a.u.)',
                    y_offset=y_offset[0]
)

# Create a subplot figure with 2 rows, 1 column
combined_fig = make_subplots(rows=2, cols=1, shared_xaxes=True, subplot_titles=[f.layout.title.text, f2.layout.title.text],
                             # row_heights=[0.5, 1],
                             vertical_spacing=0.15)

# Add traces from f2 (second plot)
for trace in f2.data:
    combined_fig.add_trace(trace, row=1, col=1)

# Add traces from f (first plot)
for trace in f.data:
    combined_fig.add_trace(trace, row=2, col=1)

# Update layout
combined_fig.update_layout(height=fig_h*2, width=fig_w, showlegend=True)
combined_fig.update_xaxes(title_text=f2.layout.xaxis.title.text, row=2, col=1)
combined_fig.update_yaxes(title_text="Intensity (a.u.)", row=1, col=1)
combined_fig.update_yaxes(title_text="Intensity (a.u.)", row=2, col=1)

combined_fig.update_layout(
    title_text="Comparison of SERS and Raman Spectra",
    template="plotly_white"
)

apc.plotly.style_plot(combined_fig, monospaced_axes="all")

# combined_fig.write_image("plots/fig2-plots.pdf")
combined_fig.show()

```
